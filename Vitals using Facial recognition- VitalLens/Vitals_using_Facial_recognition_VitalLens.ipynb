```python
%%writefile app2.py
import streamlit as st
import cv2
import numpy as np
import vitallens
import tempfile
import matplotlib.pyplot as plt

st.set_page_config(page_title="VitalLens Video Analysis II", layout="wide")
st.title("VitalLens Video Analysis App")

# ==========================
# Step 1: Video Input
# ==========================
st.write("Upload a video file or provide a local path for analysis.")
video_file = st.file_uploader("Upload Video", type=["mp4", "avi", "mov"])
video_path_input = st.text_input("Or enter local video path:")

video_path = None
if video_file is not None:
    tfile = tempfile.NamedTemporaryFile(delete=False)
    tfile.write(video_file.read())
    video_path = tfile.name
elif video_path_input:
    video_path = video_path_input

if video_path:
    st.video(video_path)

    if st.button("Analyze Video"):
        with st.spinner("Loading video..."):
            # Load video frames
            cap = cv2.VideoCapture(video_path)
            fps = cap.get(cv2.CAP_PROP_FPS) or 30  # fallback if FPS not detected
            frames = []
            while True:
                ret, frame = cap.read()
                if not ret:
                    break
                frames.append(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
            cap.release()
            video_array = np.array(frames)

        st.success(f"Video loaded: {len(frames)} frames at {fps:.1f} FPS")
        st.info(f"Duration: {len(frames)/fps:.2f} seconds, Shape: {video_array.shape}")

        # ==========================
        # Step 2: Initialize VitalLens
        # ==========================
        API_KEY = "GdsxHv55ys3pWHUEc8tZm5Sr6H5oeb7n30NouKyX"  # Replace with your actual key
        vl = vitallens.VitalLens(
            method=vitallens.Method.VITALLENS,
            api_key=API_KEY,
            mode=vitallens.Mode.BURST,
            export_to_json=True,
            export_dir=tempfile.gettempdir(),
            estimate_rolling_vitals=True
        )

        # ==========================
        # Step 3: Analyze Video
        # ==========================
        with st.spinner("Analyzing video with VitalLens..."):
            results = vl(video_array, fps=fps, export_filename="vitals_analysis")

        if not results:
            st.warning("No results found - no face detected!")
        else:
            vital_signs = results[0]['vital_signs']

            hr_global = vital_signs.get('heart_rate', {}).get('value')
            rr_global = vital_signs.get('respiratory_rate', {}).get('value')
            has_rolling_hr = 'rolling_heart_rate' in vital_signs
            has_rolling_rr = 'rolling_respiratory_rate' in vital_signs

            st.subheader("Global Estimates")
            if hr_global: st.write(f"**Heart Rate:** {hr_global:.1f} bpm")
            if rr_global: st.write(f"**Respiratory Rate:** {rr_global:.1f} rpm")

            # ==========================
            # Step 4: Plot Rolling Vital Signs
            # ==========================
            if has_rolling_hr:
                rolling_hr = vital_signs['rolling_heart_rate']['data']
                time_axis = np.arange(len(rolling_hr)) / fps
                plt.figure(figsize=(10,4))
                plt.plot(time_axis, rolling_hr, color='#ef4444', label='Heart Rate')
                if hr_global:
                    plt.axhline(y=hr_global, color='red', linestyle='--', label=f'Global Avg: {hr_global:.1f}')
                plt.xlabel("Time (s)")
                plt.ylabel("Heart Rate (bpm)")
                plt.title("Rolling Heart Rate")
                plt.grid(True)
                plt.legend()
                st.pyplot(plt.gcf())
                plt.close()

            if has_rolling_rr:
                rolling_rr = vital_signs['rolling_respiratory_rate']['data']
                time_axis = np.arange(len(rolling_rr)) / fps
                plt.figure(figsize=(10,4))
                plt.plot(time_axis, rolling_rr, color='#3b82f6', label='Respiratory Rate')
                if rr_global:
                    plt.axhline(y=rr_global, color='blue', linestyle='--', label=f'Global Avg: {rr_global:.1f}')
                plt.xlabel("Time (s)")
                plt.ylabel("Respiratory Rate (rpm)")
                plt.title("Rolling Respiratory Rate")
                plt.grid(True)
                plt.legend()
                st.pyplot(plt.gcf())
                plt.close()

            # Combined normalized plot if both available
            if has_rolling_hr and has_rolling_rr:
                hr_norm = (rolling_hr - np.nanmin(rolling_hr)) / (np.nanmax(rolling_hr) - np.nanmin(rolling_hr))
                rr_norm = (rolling_rr - np.nanmin(rolling_rr)) / (np.nanmax(rolling_rr) - np.nanmin(rolling_rr))
                plt.figure(figsize=(10,4))
                plt.plot(time_axis, hr_norm, color='#ef4444', label='Heart Rate (normalized)')
                plt.plot(time_axis, rr_norm, color='#3b82f6', label='Respiratory Rate (normalized)')
                plt.xlabel("Time (s)")
                plt.ylabel("Normalized Value (0-1)")
                plt.title("Normalized Vital Signs Comparison")
                plt.grid(True)
                plt.legend()
                st.pyplot(plt.gcf())
                plt.close()

            st.success("Analysis complete!")
            st.info(f"JSON results saved at: {tempfile.gettempdir()}/vitals_analysis.json")
```
