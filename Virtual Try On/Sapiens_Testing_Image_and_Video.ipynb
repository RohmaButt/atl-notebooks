```python
import subprocess
import os

def download_youtube_video(url, output_path=".", resolution="best", start_time="00:00:16", end_time="00:00:21"):
    """
    Download a specific time segment of a YouTube video using yt-dlp

    Args:
        url (str): YouTube video URL
        output_path (str): Directory to save the video
        resolution (str): Video resolution ('best', '1080p', '720p', etc.)
        start_time (str): Start time in HH:MM:SS
        end_time (str): End time in HH:MM:SS

    Returns:
        str: Path to downloaded video or error message
    """
    try:
        # Install yt-dlp if not already installed
        subprocess.run(["pip", "install", "yt-dlp"], stdout=subprocess.PIPE, stderr=subprocess.PIPE)

        # Create output directory if it doesn't exist
        os.makedirs(output_path, exist_ok=True)

        # Set resolution format
        if resolution == "best":
            format_str = "bestvideo+bestaudio/best"
        else:
            format_str = f"bestvideo[height<={resolution.rstrip('p')}]+bestaudio/best"

        # Set output template
        output_template = os.path.join(output_path, "%(title)s.%(ext)s")

        # Set the download section string (e.g., "*16-21")
        section = f"*{start_time}-{end_time}"

        # Command to execute
        command = [
            "yt-dlp",
            "-f", format_str,
            "--download-sections", section,
            "--merge-output-format", "mp4",
            "-o", output_template,
            url
        ]

        print(f"Downloading segment {start_time} to {end_time} from: {url}")
        print(f"Selected resolution: {resolution}")

        # Execute the command
        result = subprocess.run(command, capture_output=True, text=True)

        if result.returncode == 0:
            # Try to find saved file in stdout
            for line in result.stdout.split('\n'):
                if "Destination" in line and ".mp4" in line:
                    filename = line.split("Destination: ")[1].strip()
                    return f"Download complete! Saved to: {filename}"
            return "Download complete! Check your output directory."
        else:
            return f"Error downloading video:\n{result.stderr}"

    except Exception as e:
        return f"An error occurred: {str(e)}"

# Example usage
video_url = "https://www.youtube.com/watch?v=S-tJmuMNCsQ"
print(download_youtube_video(video_url, resolution="2160"))
```

```python
import torch
import cv2
import numpy as np
import os
import sys
from IPython.display import display, HTML

# Check if we're in Google Colab
IN_COLAB = 'google.colab' in sys.modules
if IN_COLAB:
    from google.colab import files
    # Install required packages if needed
    !pip install sapiens-inference --quiet

# Import the sapiens libraries with error handling
try:
    from sapiens_inference.normal import SapiensNormal, SapiensNormalType
    from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType
    print("Sapiens libraries successfully imported")
except ImportError as e:
    print(f"Error importing Sapiens libraries: {e}")
    print("Please make sure sapiens-inference is installed: pip install sapiens-inference")
    sys.exit(1)

def warp_logo_with_normals(logo, normal_map, mask):
    # Convert normal map values from [-1,1] to displacement values
    normal_h, normal_w = normal_map.shape[:2]

    # Create meshgrid for warping
    rows, cols = normal_h, normal_w
    map_x, map_y = np.meshgrid(np.arange(cols), np.arange(rows))

    # Extract normal map components and scale them appropriately
    nx = normal_map[:, :, 0]  # X component
    ny = normal_map[:, :, 1]  # Y component
    nz = normal_map[:, :, 2]  # Z component

    # Calculate displacement based on surface orientation
    strength = 20.0  # Adjust this value to control warping strength
    dx = nx * strength * (1.0 - nz)
    dy = ny * strength * (1.0 - nz)

    # Apply displacement with perspective correction
    map_x = map_x + dx
    map_y = map_y + dy

    # Ensure coordinates are within bounds
    map_x = np.clip(map_x, 0, cols - 1).astype(np.float32)
    map_y = np.clip(map_y, 0, rows - 1).astype(np.float32)

    # Resize logo to match the ROI size
    logo_resized = cv2.resize(logo, (normal_w, normal_h))

    # Warp the logo
    warped_logo = cv2.remap(logo_resized, map_x, map_y,
                           interpolation=cv2.INTER_LINEAR,
                           borderMode=cv2.BORDER_REPLICATE)

    # Apply shading based on normal map
    light_direction = np.array([0.0, 0.0, 1.0])
    light_intensity = np.sum(normal_map * light_direction, axis=2)
    light_intensity = np.clip(light_intensity * 0.7 + 0.3, 0.2, 1.0)

    # Apply lighting to warped logo
    warped_logo = warped_logo.astype(np.float32)
    if warped_logo.shape[2] == 4:
        # Handle alpha channel separately
        alpha = warped_logo[:, :, 3] / 255.0
        rgb = warped_logo[:, :, :3]
        rgb = rgb * light_intensity[:, :, np.newaxis]
        warped_logo[:, :, :3] = rgb
        warped_logo[:, :, 3] = alpha * 255

    return warped_logo.astype(np.uint8)

# Display a sample frame with visualization for Colab
def display_sample_frame(frame, segmentation_map=None, normal_map=None):
    """Display a sample frame with optional visualizations in Colab"""
    # Create a copy to avoid modifying the original
    display_frame = frame.copy()

    # If segmentation map is provided, show upper clothing mask overlay
    if segmentation_map is not None:
        upper_clothing_mask = (segmentation_map == 22).astype(np.uint8) * 255
        # Create a colored overlay
        overlay = np.zeros_like(display_frame)
        overlay[:,:,1] = upper_clothing_mask  # Green channel
        display_frame = cv2.addWeighted(display_frame, 0.7, overlay, 0.3, 0)

    # Show the frame
    rgb_frame = cv2.cvtColor(display_frame, cv2.COLOR_BGR2RGB)

    # Convert to JPEG for display
    _, encoded_img = cv2.imencode('.jpg', rgb_frame)
    return encoded_img.tobytes()

def process_video(input_video_path, logo_path, output_video_path=None):
    # Initialize models
    dtype = torch.float16
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B, dtype=dtype)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)

    # Load logo
    logo = cv2.imread(logo_path, cv2.IMREAD_UNCHANGED)
    if logo is None:
        raise ValueError(f"Could not load logo image from {logo_path}")

    # Open the video file
    cap = cv2.VideoCapture(input_video_path)
    if not cap.isOpened():
        raise ValueError(f"Could not open video file: {input_video_path}")

    # Get video properties
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # Setup output video writer if path is provided
    output_writer = None
    if output_video_path:
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # You might need to adjust this codec based on your system
        output_writer = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    frame_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1
        print(f"Processing frame {frame_count}/{total_frames} ({frame_count/total_frames*100:.1f}%)")

        # Get segmentation mask
        segmentation_map = segmentation_estimator(frame)
        upper_clothing_mask = (segmentation_map == 22).astype(np.uint8)  # 22 is typically the ID for upper clothing

        if np.any(upper_clothing_mask):
            # Get normal map
            normal_map = normal_estimator(frame)

            # Get ROI coordinates
            y_indices, x_indices = np.where(upper_clothing_mask > 0)
            if len(y_indices) == 0 or len(x_indices) == 0:
                continue

            top, bottom = np.min(y_indices), np.max(y_indices)
            left, right = np.min(x_indices), np.max(x_indices)

            # Extract ROI and normal map for the t-shirt area
            roi = frame[top:bottom, left:right].copy()
            roi_normal = normal_map[top:bottom, left:right]
            roi_mask = upper_clothing_mask[top:bottom, left:right]

            # Warp logo according to normal map
            warped_logo = warp_logo_with_normals(logo, roi_normal, roi_mask)

            # Blend warped logo with original frame
            if warped_logo.shape[2] == 4:
                alpha = warped_logo[:, :, 3] / 255.0
                alpha = np.expand_dims(alpha, axis=2)
                alpha = np.repeat(alpha, 3, axis=2)

                blended = (1.0 - alpha) * roi + alpha * warped_logo[:, :, :3]
                frame[top:bottom, left:right] = blended.astype(np.uint8)

        # For Colab, no need to display each frame with cv2.imshow

        # Write to output video if specified
        if output_writer:
            output_writer.write(frame)

        # No need for waitKey in Colab environment

    # Release resources
    cap.release()
    if output_writer:
        output_writer.release()
    # No need for cv2.destroyAllWindows() in Colab

    print(f"Processing complete! Output saved to: {output_video_path}" if output_video_path else "Processing complete!")

def upload_files():
    """Simulate uploading video and logo files for Google Colab"""
    video_path = "demo.mp4"
    logo_path = "netflix.png"

    # Define output filename based on video file name
    output_filename = f"processed_{os.path.basename(video_path)}"

    return video_path, logo_path, output_filename

# Main execution for Colab
if __name__ == "__main__":
    print("T-shirt Logo Warping for Videos")
    print("-------------------------------")

    # Upload files
    video_path, logo_path, output_path = upload_files()

    print(f"\nProcessing video: {video_path}")
    print(f"Using logo: {logo_path}")
    print(f"Output will be saved to: {output_path}")

    # Process the video
    process_video(video_path, logo_path, output_path)

    # Provide download link for the output
    print("\nProcessing complete! Click below to download the result:")
    files.download(output_path)
```

```python
import torch
import cv2
import numpy as np
import os
from sapiens_inference.normal import SapiensNormal, SapiensNormalType, draw_normal_map
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def process_image(image_path, output_path=None, clothing_class=22, device=None, dtype=torch.float16):
    """
    Process an image to apply normal map estimation on clothing regions.

    Args:
        image_path: Path to the input image
        output_path: Path to save the output image. If None, display the image instead
        clothing_class: Segmentation class ID for clothing (default is 22 for "Upper Clothing")
        device: Device to run inference on (None for auto-detection)
        dtype: Data type for normal estimation (float16 or float32)
    """
    # Set device if not provided
    if device is None:
        device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

    print(f"Using device: {device}")
    if device.type == 'cuda':
        print(f"GPU: {torch.cuda.get_device_name(device)}")

    # Load image
    frame = cv2.imread(image_path)
    if frame is None:
        raise ValueError(f"Could not load image from {image_path}")

    # Resize if the image is too large (optional)
    max_dim = 1024
    h, w = frame.shape[:2]
    if max(h, w) > max_dim:
        scale = max_dim / max(h, w)
        frame = cv2.resize(frame, (int(w * scale), int(h * scale)))
        print(f"Resized image to {frame.shape[1]}x{frame.shape[0]}")

    # Initialize SapiensNormal estimator for surface normal estimation
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B, dtype=dtype, device=device)

    # Initialize SapiensSegmentation estimator for body-part segmentation
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B, device=device)

    # Step 1: Run segmentation to get mask for clothing
    print("Running segmentation...")
    segmentation_map = segmentation_estimator(frame)

    # Create binary mask for the specified clothing class
    clothing_mask = (segmentation_map == clothing_class).astype(np.uint8)

    # Optionally apply morphological operations to improve the mask
    kernel = np.ones((5, 5), np.uint8)
    clothing_mask = cv2.morphologyEx(clothing_mask, cv2.MORPH_CLOSE, kernel)

    # Step 2: Apply normal map estimation
    print("Estimating normal map...")
    normal_map = normal_estimator(frame)
    normal_map_visualization = draw_normal_map(normal_map)

    # Step 3: Use mask to blend normal map with original frame
    if np.any(clothing_mask):  # Check if there is any detected clothing
        print(f"Clothing detected (class {clothing_class})")

        # Apply the mask to the normal map
        masked_normal_map = cv2.bitwise_and(normal_map_visualization, normal_map_visualization,
                                            mask=clothing_mask)

        # Inverse mask for original image
        inv_mask = 1 - clothing_mask
        masked_frame = cv2.bitwise_and(frame, frame, mask=inv_mask)

        # Combine masked areas
        result = masked_frame + masked_normal_map
    else:
        print(f"No clothing of class {clothing_class} detected in the image")
        result = frame.copy()

    # Save or display the result
    if output_path:
        os.makedirs(os.path.dirname(output_path) if os.path.dirname(output_path) else '.', exist_ok=True)
        cv2.imwrite(output_path, result)
        print(f"Saved result to {output_path}")

        # Also save the mask for debugging
        mask_path = output_path.replace('.', '_mask.')
        cv2.imwrite(mask_path, clothing_mask * 255)
        print(f"Saved mask to {mask_path}")

        # Save the normal map for reference
        normal_path = output_path.replace('.', '_normal.')
        cv2.imwrite(normal_path, normal_map_visualization)
        print(f"Saved normal map to {normal_path}")
    else:
        # Display results
        cv2.imshow("Original Image", frame)
        cv2.imshow("Clothing Mask", clothing_mask * 255)
        cv2.imshow("Normal Map", normal_map_visualization)
        cv2.imshow("Result", result)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

    return result

if __name__ == "__main__":
    # Set your paths here
    input_image_path = "/content/drive/MyDrive/ATL-Project/demo.jpg"  # Path to your input image
    output_image_path = "/content/drive/MyDrive/ATL-Project/result.png"  # Path where the result will be saved

    # Set parameters
    clothing_class = 22  # 22 is typically "Upper Clothing" in most segmentation models
    use_gpu = True  # Set to False to force CPU usage
    use_half_precision = True  # Set to False for more accurate but slower processing

    # Set device
    device = torch.device('cuda' if use_gpu and torch.cuda.is_available() else 'cpu')

    # Set precision
    dtype = torch.float16 if use_half_precision else torch.float32

    # Process the image
    process_image(input_image_path, output_image_path, clothing_class, device, dtype)
```

```python
import cv2
import numpy as np
import torch
import argparse
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType
from sapiens_inference.pose import SapiensPoseEstimation, SapiensPoseEstimationType

def load_texture(texture_path):
    """Load texture with alpha channel handling"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Failed to load texture from {texture_path}")

    # Add alpha channel if missing
    if texture.shape[2] == 3:
        alpha = np.ones(texture.shape[:2], dtype=texture.dtype) * 255
        texture = cv2.merge([texture[:,:,0], texture[:,:,1], texture[:,:,2], alpha])

    return cv2.resize(texture, (512, 512), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Calculate tangent space vectors from normal map"""
    nx, ny, nz = normal_map[:,:,0], normal_map[:,:,1], normal_map[:,:,2]

    tangent = np.stack([nz, np.zeros_like(nx), -nx], axis=2)
    tangent /= (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    bitangent = np.cross(normal_map, tangent)
    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask):
    """Apply texture to masked region using normal-based warping"""
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = coords[0].min(), coords[0].max()
    x_min, x_max = coords[1].min(), coords[1].max()

    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute coordinate warping
    tangent, _ = compute_normal_based_mapping(roi_normal)
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)

    # Apply deformation
    deform_strength = 0.1
    warped_coords = np.stack([
        x_coords + tangent[:,:,0] * deform_strength,
        y_coords + tangent[:,:,1] * deform_strength
    ], axis=-1)

    # Sample texture
    tex_h, tex_w = texture.shape[:2]
    tex_x = (warped_coords[:,:,0] / roi_normal.shape[1]) * tex_w
    tex_y = (warped_coords[:,:,1] / roi_normal.shape[0]) * tex_h

    warped_texture = cv2.remap(texture,
                              tex_x.astype(np.float32),
                              tex_y.astype(np.float32),
                              cv2.INTER_LINEAR,
                              borderMode=cv2.BORDER_WRAP)

    # Lighting and blending
    lighting = np.clip(np.dot(roi_normal, [0,0,1]), 0.3, 1.0)[:,:,np.newaxis]
    alpha = (warped_texture[:,:,3] / 255.0 * roi_mask)[:,:,np.newaxis] * lighting

    frame_roi = frame[y_min:y_max, x_min:x_max]
    blended_roi = frame_roi * (1 - alpha) + warped_texture[:,:,:3] * alpha
    frame[y_min:y_max, x_min:x_max] = blended_roi.astype(np.uint8)

    return frame

def draw_body_keypoints(result_img, keypoints):
    """Draw only body keypoints (exclude face) on the image"""
    # COCO body keypoint indices (17 keypoints for body only)
    body_keypoint_indices = [
        "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
        "left_wrist", "right_wrist", "left_hip", "right_hip",
        "left_knee", "right_knee", "left_ankle", "right_ankle"
    ]

    # Iterate through detected persons
    if keypoints is not None:
        for person in keypoints:
            for kp_name, (x, y, conf) in person.items():
                if kp_name in body_keypoint_indices and conf > 0.5:  # Confidence threshold
                    if 0 <= x < result_img.shape[1] and 0 <= y < result_img.shape[0]:  # Ensure within image bounds
                        cv2.circle(result_img, (int(x), int(y)), 5, (255, 0, 0), -1)

    return result_img

def process_single_image(input_path, texture_path, output_path):
    """Process a single image with virtual try-on"""
    # Read input image
    frame = cv2.imread(input_path)
    if frame is None:
        raise ValueError(f"Failed to load image from {input_path}")

    # Load texture
    texture = load_texture(texture_path)

    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    seg_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)
    pose_estimator = SapiensPoseEstimation(
        SapiensPoseEstimationType.POSE_ESTIMATION_03B,
        dtype=torch.float16
    )

    # Process image
    segmentation_map = seg_estimator(frame)
    normal_map = normal_estimator(frame)
    shirt_mask = (segmentation_map == 22).astype(np.uint8)  # 22 is the class ID for shirt

    # Apply texture mapping
    if np.any(shirt_mask):
        frame = apply_texture_mapping(frame, texture, normal_map, shirt_mask)

    # Pose estimation
    result_img, keypoints = pose_estimator(frame)

    # Draw body keypoints
    result_img = draw_body_keypoints(result_img, keypoints)

    # Save the result
    cv2.imwrite(output_path, result_img)

    print(f"Processed image saved to {output_path}")

    # Optionally display the result
    # cv2.imshow('Virtual Try-On Result', result_img)
    # cv2.waitKey(0)
    # cv2.destroyAllWindows()

def main():
    # parser = argparse.ArgumentParser(description="Virtual Try-On for a single image")
    # parser.add_argument("--input", "-i", required=True, help="Path to input image")
    # parser.add_argument("--texture", "-t", required=True, help="Path to texture image")
    # parser.add_argument("--output", "-o", default="result.jpg", help="Path to save output image")
    # args = parser.parse_args()
    input_image_path = "/content/drive/MyDrive/ATL-Project/demo.jpg"  # Path to your input image
    output_image_path = "/content/drive/MyDrive/ATL-Project/result.png"  # Path where the result will be
    texture_image_path = "/content/drive/MyDrive/ATL-Project/netflix.png"

    process_single_image(input_image_path, texture_image_path, output_image_path)

if __name__ == "__main__":
    main()
```

```python
import cv2
import numpy as np
import torch
import argparse
import time
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType
from sapiens_inference.pose import SapiensPoseEstimation, SapiensPoseEstimationType

def load_texture(texture_path):
    """Load texture with alpha channel handling"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Failed to load texture from {texture_path}")

    # Add alpha channel if missing
    if texture.shape[2] == 3:
        alpha = np.ones(texture.shape[:2], dtype=texture.dtype) * 255
        texture = cv2.merge([texture[:,:,0], texture[:,:,1], texture[:,:,2], alpha])

    return cv2.resize(texture, (512, 512), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Calculate tangent space vectors from normal map"""
    nx, ny, nz = normal_map[:,:,0], normal_map[:,:,1], normal_map[:,:,2]

    tangent = np.stack([nz, np.zeros_like(nx), -nx], axis=2)
    tangent /= (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    bitangent = np.cross(normal_map, tangent)
    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask):
    """Apply texture to masked region using normal-based warping"""
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = coords[0].min(), coords[0].max()
    x_min, x_max = coords[1].min(), coords[1].max()

    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute coordinate warping
    tangent, _ = compute_normal_based_mapping(roi_normal)
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)

    # Apply deformation
    deform_strength = 0.1
    warped_coords = np.stack([
        x_coords + tangent[:,:,0] * deform_strength,
        y_coords + tangent[:,:,1] * deform_strength
    ], axis=-1)

    # Sample texture
    tex_h, tex_w = texture.shape[:2]
    tex_x = (warped_coords[:,:,0] / roi_normal.shape[1]) * tex_w
    tex_y = (warped_coords[:,:,1] / roi_normal.shape[0]) * tex_h

    warped_texture = cv2.remap(texture,
                              tex_x.astype(np.float32),
                              tex_y.astype(np.float32),
                              cv2.INTER_LINEAR,
                              borderMode=cv2.BORDER_WRAP)

    # Lighting and blending
    lighting = np.clip(np.dot(roi_normal, [0,0,1]), 0.3, 1.0)[:,:,np.newaxis]
    alpha = (warped_texture[:,:,3] / 255.0 * roi_mask)[:,:,np.newaxis] * lighting

    frame_roi = frame[y_min:y_max, x_min:x_max]
    blended_roi = frame_roi * (1 - alpha) + warped_texture[:,:,:3] * alpha
    frame[y_min:y_max, x_min:x_max] = blended_roi.astype(np.uint8)

    return frame

def draw_body_keypoints_and_bboxes(result_img, keypoints, bboxes):
    """Draw body keypoints and bounding boxes on the image"""
    # COCO body keypoint indices (excluding face)
    body_keypoint_indices = [
        "left_shoulder", "right_shoulder", "left_elbow", "right_elbow",
        "left_wrist", "right_wrist", "left_hip", "right_hip",
        "left_knee", "right_knee", "left_ankle", "right_ankle"
    ]

    # Draw bounding boxes with a different color
    if bboxes is not None:
        for box in bboxes:
            # Extract coordinates (assuming bbox format is [x1, y1, x2, y2])
            x1, y1, x2, y2 = map(int, box[:4])
            # Draw rectangle with green color and 2px thickness
            cv2.rectangle(result_img, (x1, y1), (x2, y2), (0, 255, 0), 2)

            # Optionally, draw confidence score if available
            if len(box) > 4:
                conf = box[4]
                cv2.putText(result_img, f"{conf:.2f}", (x1, y1-5),
                           cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 1)

    # Iterate through detected persons to draw keypoints
    if keypoints is not None:
        for person in keypoints:
            for kp_name, (x, y, conf) in person.items():
                if kp_name in body_keypoint_indices and conf > 0.5:  # Confidence threshold
                    if 0 <= x < result_img.shape[1] and 0 <= y < result_img.shape[0]:  # Ensure within image bounds
                        cv2.circle(result_img, (int(x), int(y)), 5, (255, 0, 0), -1)

    return result_img

def crop_and_pad_bbox(frame, bbox, padding=10):
    """
    Crop a person from the frame based on bbox with padding and ensure valid coordinates
    Returns the cropped image and coordinates for placing it back
    """
    img_h, img_w = frame.shape[:2]

    # Extract bbox coordinates with padding
    x1, y1, x2, y2 = map(int, bbox[:4])
    x1 = max(0, x1 - padding)
    y1 = max(0, y1 - padding)
    x2 = min(img_w, x2 + padding)
    y2 = min(img_h, y2 + padding)

    # Crop the region
    person_crop = frame[y1:y2, x1:x2].copy()

    return person_crop, (x1, y1, x2, y2)

def process_frame(frame, texture, normal_estimator, seg_estimator, pose_estimator):
    """Process a single frame with virtual try-on"""
    # First get the bounding boxes for all people
    bboxes = pose_estimator.detector.detect(frame)

    # Create a copy of the original frame for processing
    result_img = frame.copy()

    # Process each person separately
    all_keypoints = []

    for bbox in bboxes:
        # Skip low confidence detections
        if len(bbox) > 4 and bbox[4] < 0.5:
            continue

        # Crop the person with some padding
        person_crop, (x1, y1, x2, y2) = crop_and_pad_bbox(frame, bbox)

        # Skip if the crop is too small
        if person_crop.shape[0] < 10 or person_crop.shape[1] < 10:
            continue

        # Process individual person
        segmentation_map = seg_estimator(person_crop)
        normal_map = normal_estimator(person_crop)

        # 22 is the class ID for shirt
        shirt_mask = (segmentation_map == 22).astype(np.uint8)

        # Apply texture mapping if shirt is detected
        if np.any(shirt_mask):
            person_crop = apply_texture_mapping(person_crop, texture, normal_map, shirt_mask)

        # Get pose estimation for the individual
        person_result, person_keypoints = pose_estimator(person_crop)

        # Adjust keypoint coordinates to match the original image
        if person_keypoints:
            for person in person_keypoints:
                for kp_name in person:
                    x, y, conf = person[kp_name]
                    person[kp_name] = (x + x1, y + y1, conf)
                all_keypoints.append(person)

        # Place the processed crop back into the result image
        result_img[y1:y2, x1:x2] = person_crop

    # Draw body keypoints and bounding boxes on the final result
    result_img = draw_body_keypoints_and_bboxes(result_img, all_keypoints, bboxes)

    return result_img

def process_single_image(input_path, texture_path, output_path):
    """Process a single image with virtual try-on"""
    # Read input image
    frame = cv2.imread(input_path)
    if frame is None:
        raise ValueError(f"Failed to load image from {input_path}")

    # Load texture
    texture = load_texture(texture_path)

    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    seg_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)
    pose_estimator = SapiensPoseEstimation(
        SapiensPoseEstimationType.POSE_ESTIMATION_03B,
        dtype=torch.float16
    )

    # Process the frame
    result_img = process_frame(frame, texture, normal_estimator, seg_estimator, pose_estimator)

    # Save the result
    cv2.imwrite(output_path, result_img)
    print(f"Processed image saved to {output_path}")

def process_video(input_path, texture_path, output_path, start_frame=0, end_frame=None, frame_interval=1):
    """Process a video with virtual try-on"""
    # Read input video
    cap = cv2.VideoCapture(input_path)
    if not cap.isOpened():
        raise ValueError(f"Failed to open video from {input_path}")

    # Get video properties
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    # Set end frame if not specified
    if end_frame is None or end_frame > frame_count:
        end_frame = frame_count

    # Load texture
    texture = load_texture(texture_path)

    # Initialize models (only once for efficiency)
    print("Initializing models...")
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    seg_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)
    pose_estimator = SapiensPoseEstimation(
        SapiensPoseEstimationType.POSE_ESTIMATION_03B,
        dtype=torch.float16
    )

    # Initialize video writer
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')  # You can change codec as needed
    out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

    # Process frames
    frame_idx = 0
    processed_count = 0

    print(f"Processing video: {input_path}")
    print(f"Total frames to process: {(end_frame - start_frame) // frame_interval}")

    start_time = time.time()

    try:
        while cap.isOpened():
            ret, frame = cap.read()
            if not ret or frame_idx >= end_frame:
                break

            # Skip frames before start_frame
            if frame_idx < start_frame:
                frame_idx += 1
                continue

            # Process at specified interval
            if (frame_idx - start_frame) % frame_interval == 0:
                print(f"Processing frame {frame_idx}/{frame_count}", end='\r')

                # Process the current frame
                try:
                    result_frame = process_frame(frame, texture, normal_estimator, seg_estimator, pose_estimator)
                    out.write(result_frame)
                    processed_count += 1

                    # Optional: save intermediate frame for debugging
                    # if processed_count % 10 == 0:
                    #    cv2.imwrite(f"debug_frame_{frame_idx}.jpg", result_frame)

                except Exception as e:
                    print(f"\nError processing frame {frame_idx}: {str(e)}")
                    # Write original frame on error
                    out.write(frame)
            else:
                # Skip processing but still write the original frame
                # Comment the next line if you want to skip non-processed frames
                out.write(frame)

            frame_idx += 1

            # Show progress every 10 frames
            if processed_count % 10 == 0:
                elapsed = time.time() - start_time
                frames_remaining = ((end_frame - frame_idx) // frame_interval) + 1
                if processed_count > 0 and elapsed > 0:
                    fps_processing = processed_count / elapsed
                    eta_seconds = frames_remaining / fps_processing if fps_processing > 0 else 0
                    eta_min = eta_seconds / 60
                    print(f"Processing speed: {fps_processing:.2f} fps, ETA: {eta_min:.1f} minutes", end='\r')

    except KeyboardInterrupt:
        print("\nProcessing interrupted by user.")

    finally:
        # Release resources
        elapsed_time = time.time() - start_time
        print(f"\nProcessed {processed_count} frames in {elapsed_time:.2f} seconds.")
        print(f"Average processing speed: {processed_count / elapsed_time:.2f} fps")

        cap.release()
        out.release()
        print(f"Video saved to {output_path}")

def main():
    # === Replace these paths with actual values ===
    input_path = "/content/drive/MyDrive/ATL-Project/run_close.mp4"         # or .mp4 for video
    texture_path = "netflix.png"
    output_path = "/content/drive/MyDrive/ATL-Project-SM/result.mp4"       # or .mp4 for video
    mode = "video"                           # or "video"

    # Video-specific parameters
    start_frame = 0
    end_frame = 1
    frame_interval = 1

    # Process based on selected mode
    if mode == "image":
        print(f"Processing image: {input_path}")
        process_single_image(input_path, texture_path, output_path)
    else:  # video mode
        print(f"Processing video: {input_path}")
        process_video(input_path, texture_path, output_path,
                      start_frame=start_frame,
                      end_frame=end_frame,
                      frame_interval=frame_interval)

    print("Processing completed successfully!")

if __name__ == "__main__":
    main()
```

```python
import cv2
import numpy as np
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def load_texture(texture_path):
    """Load checkerboard texture with proper alpha channel"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Could not load texture from {texture_path}")

    if len(texture.shape) == 3 and texture.shape[2] == 3:
        bgr = texture
        alpha = np.ones(texture.shape[:2], dtype=bgr.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])

    target_size = 512
    return cv2.resize(texture, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Compute texture mapping based on surface normals"""
    # Extract normal components
    nx = normal_map[:,:,0]
    ny = normal_map[:,:,1]
    nz = normal_map[:,:,2]

    # Compute tangent space basis vectors
    tangent = np.zeros_like(normal_map)
    tangent[:,:,0] = nz
    tangent[:,:,1] = 0
    tangent[:,:,2] = -nx
    tangent = tangent / (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    # Compute bitangent vectors
    bitangent = np.cross(normal_map, tangent)

    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask):
    """Apply texture with proper normal-based mapping"""
    h, w = frame.shape[:2]
    texture_h, texture_w = texture.shape[:2]

    # Find shirt region
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = np.min(coords[0]), np.max(coords[0])
    x_min, x_max = np.min(coords[1]), np.max(coords[1])

    # Get ROI
    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute normal-based mapping
    tangent, bitangent = compute_normal_based_mapping(roi_normal)

    # Create coordinate grid
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)
    coords = np.stack([x_coords, y_coords], axis=-1)

    # Apply normal-based deformation
    deform_strength = 0.1
    coords_3d = np.stack([
        coords[:,:,0] + tangent[:,:,0] * deform_strength,
        coords[:,:,1] + tangent[:,:,1] * deform_strength,
        np.ones_like(coords[:,:,0]) + tangent[:,:,2] * deform_strength
    ], axis=-1)

    # Project to texture space
    tex_x = ((coords_3d[:,:,0] / coords_3d[:,:,2]) / roi_normal.shape[1] * texture_w)
    tex_y = ((coords_3d[:,:,1] / coords_3d[:,:,2]) / roi_normal.shape[0] * texture_h)

    # Sample texture
    warped_texture = cv2.remap(texture,
                             np.clip(tex_x, 0, texture_w-1).astype(np.float32),
                             np.clip(tex_y, 0, texture_h-1).astype(np.float32),
                             cv2.INTER_LINEAR,
                             borderMode=cv2.BORDER_WRAP)

    # Calculate lighting
    light_dir = np.array([0.0, 0.0, 1.0])
    lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.3, 1.0)
    lighting = np.expand_dims(lighting, axis=2)

    # Apply alpha blending with lighting
    alpha = warped_texture[:,:,3] / 255.0 * roi_mask
    alpha = np.expand_dims(alpha, axis=2) * lighting

    # Blend with frame
    roi = frame[y_min:y_max, x_min:x_max]
    roi_new = roi * (1 - alpha) + warped_texture[:,:,:3] * alpha
    frame[y_min:y_max, x_min:x_max] = roi_new.astype(np.uint8)

    return frame

def main():
    # Define direct paths for input and output
    input_image_path = '/content/drive/MyDrive/ATL-Project/test.jpg'  # Replace with actual image path
    texture_path = '/content/drive/MyDrive/ATL-Project/fabric_texture_brown.jpg'  # Replace with actual texture path
    output_image_path = '/content/drive/MyDrive/ATL-Project/result.png'  # Replace with desired output path

    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)

    # Load input image
    frame = cv2.imread(input_image_path)
    if frame is None:
        print(f"Error: Could not load image from {input_image_path}")
        return

    # Load texture
    texture = load_texture(texture_path)

    # Process the image
    segmentation_map = segmentation_estimator(frame)
    normal_map = normal_estimator(frame)
    mask = (segmentation_map == 22).astype(np.uint8)  # 22 corresponds to the upper clothes category

    # Apply texture mapping if shirt is detected
    if np.any(mask):
        result = apply_texture_mapping(frame, texture, normal_map, mask)

        # Save the result
        cv2.imwrite(output_image_path, result)
        print(f"Result saved to {output_image_path}")

        # Optionally display the result
        cv2.imshow('Virtual Try-On Result', result)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("No shirt detected in the image")

if __name__ == "__main__":
    main()
```

```python
# Clear RAM and GPU memory before starting
import gc
import torch

gc.collect()
if torch.cuda.is_available():
    torch.cuda.empty_cache()
    torch.cuda.synchronize()
print("✓ Memory cleared!")

# Import required libraries
import cv2
import numpy as np
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def load_texture(texture_path):
    """Load checkerboard texture with proper alpha channel"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Could not load texture from {texture_path}")

    if len(texture.shape) == 3 and texture.shape[2] == 3:
        bgr = texture
        alpha = np.ones(texture.shape[:2], dtype=bgr.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])

    target_size = 512
    return cv2.resize(texture, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Compute texture mapping based on surface normals"""
    # Extract normal components
    nx = normal_map[:,:,0]
    ny = normal_map[:,:,1]
    nz = normal_map[:,:,2]

    # Compute tangent space basis vectors
    tangent = np.zeros_like(normal_map)
    tangent[:,:,0] = nz
    tangent[:,:,1] = 0
    tangent[:,:,2] = -nx
    tangent = tangent / (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    # Compute bitangent vectors
    bitangent = np.cross(normal_map, tangent)

    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask, opacity=1.0, deform_strength=0.1, lighting_intensity=0.0):
    """Apply texture with proper normal-based mapping

    Args:
        frame: Input image
        texture: Texture to apply
        normal_map: Surface normals
        mask: Segmentation mask
        opacity: Float between 0.0-1.0. 1.0 = full opaque, 0.5 = half transparent
        deform_strength: How much texture warps with surface (0.0 = flat, 0.5 = follows curves)
        lighting_intensity: How much shading to apply (0.0 = no shading/original colors, 1.0 = full shading)
    """
    h, w = frame.shape[:2]
    texture_h, texture_w = texture.shape[:2]

    # Find shirt region
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = np.min(coords[0]), np.max(coords[0])
    x_min, x_max = np.min(coords[1]), np.max(coords[1])

    # Get ROI
    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute normal-based mapping
    tangent, bitangent = compute_normal_based_mapping(roi_normal)

    # Create coordinate grid
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)
    coords = np.stack([x_coords, y_coords], axis=-1)

    # Apply normal-based deformation
    coords_3d = np.stack([
        coords[:,:,0] + tangent[:,:,0] * deform_strength,
        coords[:,:,1] + tangent[:,:,1] * deform_strength,
        np.ones_like(coords[:,:,0]) + tangent[:,:,2] * deform_strength
    ], axis=-1)

    # Project to texture space
    tex_x = ((coords_3d[:,:,0] / coords_3d[:,:,2]) / roi_normal.shape[1] * texture_w)
    tex_y = ((coords_3d[:,:,1] / coords_3d[:,:,2]) / roi_normal.shape[0] * texture_h)

    # Sample texture
    warped_texture = cv2.remap(texture,
                             np.clip(tex_x, 0, texture_w-1).astype(np.float32),
                             np.clip(tex_y, 0, texture_h-1).astype(np.float32),
                             cv2.INTER_LINEAR,
                             borderMode=cv2.BORDER_WRAP)

    # Get base texture colors
    textured_color = warped_texture[:,:,:3].astype(np.float32)

    # Apply lighting if intensity > 0
    if lighting_intensity > 0:
        light_dir = np.array([0.0, 0.0, 1.0])
        # Lighting range: lerp between (1.0, 1.0) and (0.3, 1.0) based on intensity
        min_lighting = 1.0 - (0.7 * lighting_intensity)  # 1.0 when intensity=0, 0.3 when intensity=1
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), min_lighting, 1.0)
        lighting = np.expand_dims(lighting, axis=2)
        textured_color = textured_color * lighting

    # Apply opacity parameter (1.0 = full opaque)
    alpha = np.expand_dims(roi_mask.astype(np.float32), axis=2) * opacity

    # Blend with frame
    roi = frame[y_min:y_max, x_min:x_max]
    roi_new = roi * (1 - alpha) + textured_color * alpha
    frame[y_min:y_max, x_min:x_max] = roi_new.astype(np.uint8)

    return frame

def main():
    # ========== CONFIGURATION PARAMETERS ==========
    # Adjust these values as needed
    OPACITY = 1.0              # 1.0 = full opaque, 0.5 = half transparent
    DEFORM_STRENGTH = 0.1      # 0.1 = subtle warping, 0.5 = follows curves more
    LIGHTING_INTENSITY = 0.0   # 0.0 = original colors (no shading), 1.0 = full shading

    # Define direct paths for input and output
    input_image_path = '/content/drive/MyDrive/ATL-Project/IMG_2989.JPEG'
    texture_path = '/content/drive/MyDrive/ATL-Project/crochet_texture.jpg'
    output_image_path = '/content/drive/MyDrive/ATL-Project/result.png'
    # ==============================================

    print("Initializing models...")
    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)

    print("Loading input image...")
    # Load input image
    frame = cv2.imread(input_image_path)
    if frame is None:
        print(f"Error: Could not load image from {input_image_path}")
        return

    print("Loading texture...")
    # Load texture
    texture = load_texture(texture_path)

    print("Processing image (this may take a moment)...")
    # Process the image
    segmentation_map = segmentation_estimator(frame)
    normal_map = normal_estimator(frame)
    mask = (segmentation_map == 22).astype(np.uint8)  # 22 corresponds to the upper clothes category

    # Apply texture mapping if shirt is detected
    if np.any(mask):
        print("Shirt detected! Applying texture...")
        result = apply_texture_mapping(
            frame,
            texture,
            normal_map,
            mask,
            opacity=OPACITY,
            deform_strength=DEFORM_STRENGTH,
            lighting_intensity=LIGHTING_INTENSITY
        )

        # Save the result
        cv2.imwrite(output_image_path, result)
        print(f"✓ Result saved to {output_image_path}")

        # Optionally display the result
        try:
            cv2.imshow('Virtual Try-On Result', result)
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        except:
            print("(Display not available in this environment)")
    else:
        print("No shirt detected in the image")

    # Clear memory after processing
    del frame, texture, segmentation_map, normal_map, mask
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    print("✓ Memory cleared after processing")

if __name__ == "__main__":
    main()
```

# Better lower clothing + upper

```python
import cv2
import numpy as np
import gc
import torch
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def load_texture(texture_path):
    """Load checkerboard texture with proper alpha channel"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Could not load texture from {texture_path}")

    if len(texture.shape) == 3 and texture.shape[2] == 3:
        bgr = texture
        alpha = np.ones(texture.shape[:2], dtype=bgr.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])
    elif len(texture.shape) == 2:
        # Grayscale image - convert to BGRA
        gray = texture
        bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
        alpha = np.ones(texture.shape[:2], dtype=gray.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])

    target_size = 512
    return cv2.resize(texture, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Compute texture mapping based on surface normals"""
    # Extract normal components
    nx = normal_map[:,:,0]
    ny = normal_map[:,:,1]
    nz = normal_map[:,:,2]

    # Compute tangent space basis vectors
    tangent = np.zeros_like(normal_map)
    tangent[:,:,0] = nz
    tangent[:,:,1] = 0
    tangent[:,:,2] = -nx
    tangent = tangent / (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    # Compute bitangent vectors
    bitangent = np.cross(normal_map, tangent)

    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask, clothing_type='upper'):
    """Apply texture with proper normal-based mapping"""
    h, w = frame.shape[:2]
    texture_h, texture_w = texture.shape[:2]

    # Find clothing region
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = np.min(coords[0]), np.max(coords[0])
    x_min, x_max = np.min(coords[1]), np.max(coords[1])

    # Get ROI
    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute normal-based mapping
    tangent, bitangent = compute_normal_based_mapping(roi_normal)

    # Create coordinate grid
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)
    coords = np.stack([x_coords, y_coords], axis=-1)

    # Apply normal-based deformation
    deform_strength = 0.1
    coords_3d = np.stack([
        coords[:,:,0] + tangent[:,:,0] * deform_strength,
        coords[:,:,1] + tangent[:,:,1] * deform_strength,
        np.ones_like(coords[:,:,0]) + tangent[:,:,2] * deform_strength
    ], axis=-1)

    # Project to texture space
    tex_x = ((coords_3d[:,:,0] / coords_3d[:,:,2]) / roi_normal.shape[1] * texture_w)
    tex_y = ((coords_3d[:,:,1] / coords_3d[:,:,2]) / roi_normal.shape[0] * texture_h)

    # Sample texture
    warped_texture = cv2.remap(texture,
                             np.clip(tex_x, 0, texture_w-1).astype(np.float32),
                             np.clip(tex_y, 0, texture_h-1).astype(np.float32),
                             cv2.INTER_LINEAR,
                             borderMode=cv2.BORDER_WRAP)

    # Safety check: ensure warped_texture has correct shape
    if len(warped_texture.shape) != 3 or warped_texture.shape[2] != 4:
        print(f"Warning: warped_texture has unexpected shape {warped_texture.shape}, reconstructing...")
        # Reconstruct with alpha channel
        if len(warped_texture.shape) == 2:
            # Grayscale
            gray = warped_texture
            bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([bgr, alpha])
        elif warped_texture.shape[2] == 3:
            # BGR without alpha
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([warped_texture, alpha])

    # Calculate lighting - Different settings for upper vs lower clothing
    light_dir = np.array([0.0, 0.0, 1.0])

    if clothing_type == 'upper':
        # Upper clothing: brighter, preserves color better
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.5, 1.0)
        lighting = np.expand_dims(lighting, axis=2)

        # Apply lighting to texture color
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2)
        textured_rgb = warped_texture[:,:,:3].astype(np.float32) * lighting

    else:  # clothing_type == 'lower'
        # Lower clothing: original method with more contrast
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.3, 1.0)
        lighting = np.expand_dims(lighting, axis=2)

        # Original blending method
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2) * lighting
        textured_rgb = warped_texture[:,:,:3].astype(np.float32)

    # Blend with frame
    roi = frame[y_min:y_max, x_min:x_max].astype(np.float32)
    roi_new = roi * (1 - alpha) + textured_rgb * alpha
    frame[y_min:y_max, x_min:x_max] = np.clip(roi_new, 0, 255).astype(np.uint8)

    return frame

def clear_memory():
    """Clear RAM and GPU memory"""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    print("✓ Memory cleared")

def main():
    # Clear memory before processing
    print("Clearing memory before processing...")
    clear_memory()

    # Define direct paths for input and output
    input_image_path = '/content/drive/MyDrive/ATL-Project/sample4.png'  # Replace with actual image path
    upper_texture_path = '/content/drive/MyDrive/ATL-Project/maroon.jpg'  # Replace with actual texture path for shirt
    lower_texture_path = '/content/drive/MyDrive/ATL-Project/black_cotton.jpg'  # Replace with actual texture path for pants
    output_image_path = '/content/drive/MyDrive/ATL-Project/result.png'  # Replace with desired output path

    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)

    # Load input image
    frame = cv2.imread(input_image_path)
    if frame is None:
        print(f"Error: Could not load image from {input_image_path}")
        return

    # Load textures
    try:
        upper_texture = load_texture(upper_texture_path)
        print("Upper texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")
        upper_texture = None

    try:
        lower_texture = load_texture(lower_texture_path)
        print("Lower texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")
        lower_texture = None

    # Process the image
    segmentation_map = segmentation_estimator(frame)
    normal_map = normal_estimator(frame)

    # Create masks for upper and lower clothing
    upper_mask = (segmentation_map == 22).astype(np.uint8)  # 22 = upper clothes
    lower_mask = (segmentation_map == 12).astype(np.uint8)  # 12 = lower clothes/pants

    result = frame.copy()

    # Apply texture mapping for upper clothing (shirt) if detected
    if np.any(upper_mask) and upper_texture is not None:
        print("Applying texture to upper clothing...")
        result = apply_texture_mapping(result, upper_texture, normal_map, upper_mask, clothing_type='upper')
    elif np.any(upper_mask):
        print("Upper clothing detected but no texture provided")
    else:
        print("No upper clothing detected in the image")

    # Apply texture mapping for lower clothing (pants) if detected
    if np.any(lower_mask) and lower_texture is not None:
        print("Applying texture to lower clothing...")
        result = apply_texture_mapping(result, lower_texture, normal_map, lower_mask, clothing_type='lower')
    elif np.any(lower_mask):
        print("Lower clothing detected but no texture provided")
    else:
        print("No lower clothing detected in the image")

    # Save the result if any texture was applied
    if not np.array_equal(result, frame):
        cv2.imwrite(output_image_path, result)
        print(f"Result saved to {output_image_path}")

        # Optionally display the result
        cv2.imshow('Virtual Try-On Result', result)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("No clothing detected or no textures applied")

    # Clear memory after processing
    print("\nClearing memory after processing...")
    del frame, result, segmentation_map, normal_map, upper_mask, lower_mask
    if upper_texture is not None:
        del upper_texture
    if lower_texture is not None:
        del lower_texture
    clear_memory()
    print("✓ Processing complete")

if __name__ == "__main__":
    main()
```

```python
import cv2
import numpy as np
import gc
import torch
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def load_texture(texture_path):
    """Load checkerboard texture with proper alpha channel"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Could not load texture from {texture_path}")

    if len(texture.shape) == 3 and texture.shape[2] == 3:
        bgr = texture
        alpha = np.ones(texture.shape[:2], dtype=bgr.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])
    elif len(texture.shape) == 2:
        # Grayscale image - convert to BGRA
        gray = texture
        bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
        alpha = np.ones(texture.shape[:2], dtype=gray.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])

    target_size = 512
    return cv2.resize(texture, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Compute texture mapping based on surface normals"""
    # Extract normal components
    nx = normal_map[:,:,0]
    ny = normal_map[:,:,1]
    nz = normal_map[:,:,2]

    # Compute tangent space basis vectors
    tangent = np.zeros_like(normal_map)
    tangent[:,:,0] = nz
    tangent[:,:,1] = 0
    tangent[:,:,2] = -nx
    tangent = tangent / (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    # Compute bitangent vectors
    bitangent = np.cross(normal_map, tangent)

    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask, clothing_type='upper'):
    """Apply texture with proper normal-based mapping"""
    h, w = frame.shape[:2]
    texture_h, texture_w = texture.shape[:2]

    # Find clothing region
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = np.min(coords[0]), np.max(coords[0])
    x_min, x_max = np.min(coords[1]), np.max(coords[1])

    # Get ROI
    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute normal-based mapping
    tangent, bitangent = compute_normal_based_mapping(roi_normal)

    # Create coordinate grid
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)
    coords = np.stack([x_coords, y_coords], axis=-1)

    # Apply normal-based deformation
    deform_strength = 0.1
    coords_3d = np.stack([
        coords[:,:,0] + tangent[:,:,0] * deform_strength,
        coords[:,:,1] + tangent[:,:,1] * deform_strength,
        np.ones_like(coords[:,:,0]) + tangent[:,:,2] * deform_strength
    ], axis=-1)

    # Project to texture space
    tex_x = ((coords_3d[:,:,0] / coords_3d[:,:,2]) / roi_normal.shape[1] * texture_w)
    tex_y = ((coords_3d[:,:,1] / coords_3d[:,:,2]) / roi_normal.shape[0] * texture_h)

    # Sample texture
    warped_texture = cv2.remap(texture,
                             np.clip(tex_x, 0, texture_w-1).astype(np.float32),
                             np.clip(tex_y, 0, texture_h-1).astype(np.float32),
                             cv2.INTER_LINEAR,
                             borderMode=cv2.BORDER_WRAP)

    # Safety check: ensure warped_texture has correct shape
    if len(warped_texture.shape) != 3 or warped_texture.shape[2] != 4:
        print(f"Warning: warped_texture has unexpected shape {warped_texture.shape}, reconstructing...")
        # Reconstruct with alpha channel
        if len(warped_texture.shape) == 2:
            # Grayscale
            gray = warped_texture
            bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([bgr, alpha])
        elif warped_texture.shape[2] == 3:
            # BGR without alpha
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([warped_texture, alpha])

    # Calculate lighting - Different settings for upper vs lower clothing
    light_dir = np.array([0.0, 0.0, 1.0])

    if clothing_type == 'upper':
        # Upper clothing: brighter, preserves color better
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.5, 1.0)
        lighting = np.expand_dims(lighting, axis=2)

        # Apply lighting to texture color
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2)
        textured_rgb = warped_texture[:,:,:3].astype(np.float32) * lighting

    else:  # clothing_type == 'lower'
        # Lower clothing: improved color preservation with better lighting
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.5, 1.0)  # Changed from 0.3 to 0.5
        lighting = np.expand_dims(lighting, axis=2)

        # Apply lighting to texture RGB instead of alpha
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2)
        textured_rgb = warped_texture[:,:,:3].astype(np.float32) * lighting

    # Blend with frame
    roi = frame[y_min:y_max, x_min:x_max].astype(np.float32)
    roi_new = roi * (1 - alpha) + textured_rgb * alpha
    frame[y_min:y_max, x_min:x_max] = np.clip(roi_new, 0, 255).astype(np.uint8)

    return frame

def clear_memory():
    """Clear RAM and GPU memory"""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()
    print("✓ Memory cleared")

def main():
    # Clear memory before processing
    print("Clearing memory before processing...")
    clear_memory()

    # Define direct paths for input and output
    input_image_path = '/content/drive/MyDrive/ATL-Project/sample4.png'  # Replace with actual image path
    upper_texture_path = '/content/drive/MyDrive/ATL-Project/maroon.jpg'  # Replace with actual texture path for shirt
    lower_texture_path = '/content/drive/MyDrive/ATL-Project/black_cotton.jpg'  # Replace with actual texture path for pants
    output_image_path = '/content/drive/MyDrive/ATL-Project/result.png'  # Replace with desired output path

    # Initialize models
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)

    # Load input image
    frame = cv2.imread(input_image_path)
    if frame is None:
        print(f"Error: Could not load image from {input_image_path}")
        return

    # Load textures
    try:
        upper_texture = load_texture(upper_texture_path)
        print("Upper texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")
        upper_texture = None

    try:
        lower_texture = load_texture(lower_texture_path)
        print("Lower texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")
        lower_texture = None

    # Process the image
    segmentation_map = segmentation_estimator(frame)
    normal_map = normal_estimator(frame)

    # Create masks for upper and lower clothing
    upper_mask = (segmentation_map == 22).astype(np.uint8)  # 22 = upper clothes
    lower_mask = (segmentation_map == 12).astype(np.uint8)  # 12 = lower clothes/pants

    result = frame.copy()

    # Apply texture mapping for upper clothing (shirt) if detected
    if np.any(upper_mask) and upper_texture is not None:
        print("Applying texture to upper clothing...")
        result = apply_texture_mapping(result, upper_texture, normal_map, upper_mask, clothing_type='upper')
    elif np.any(upper_mask):
        print("Upper clothing detected but no texture provided")
    else:
        print("No upper clothing detected in the image")

    # Apply texture mapping for lower clothing (pants) if detected
    if np.any(lower_mask) and lower_texture is not None:
        print("Applying texture to lower clothing...")
        result = apply_texture_mapping(result, lower_texture, normal_map, lower_mask, clothing_type='lower')
    elif np.any(lower_mask):
        print("Lower clothing detected but no texture provided")
    else:
        print("No lower clothing detected in the image")

    # Save the result if any texture was applied
    if not np.array_equal(result, frame):
        cv2.imwrite(output_image_path, result)
        print(f"Result saved to {output_image_path}")

        # Optionally display the result
        cv2.imshow('Virtual Try-On Result', result)
        cv2.waitKey(0)
        cv2.destroyAllWindows()
    else:
        print("No clothing detected or no textures applied")

    # Clear memory after processing
    print("\nClearing memory after processing...")
    del frame, result, segmentation_map, normal_map, upper_mask, lower_mask
    if upper_texture is not None:
        del upper_texture
    if lower_texture is not None:
        del lower_texture
    clear_memory()
    print("✓ Processing complete")

if __name__ == "__main__":
    main()
```

```python
import cv2
import numpy as np
import gc
import torch
import os
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

def load_texture(texture_path):
    """Load checkerboard texture with proper alpha channel"""
    texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
    if texture is None:
        raise ValueError(f"Could not load texture from {texture_path}")

    if len(texture.shape) == 3 and texture.shape[2] == 3:
        bgr = texture
        alpha = np.ones(texture.shape[:2], dtype=bgr.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])
    elif len(texture.shape) == 2:
        # Grayscale image - convert to BGRA
        gray = texture
        bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
        alpha = np.ones(texture.shape[:2], dtype=gray.dtype) * 255
        texture = cv2.merge([bgr[:,:,0], bgr[:,:,1], bgr[:,:,2], alpha])

    target_size = 512
    return cv2.resize(texture, (target_size, target_size), interpolation=cv2.INTER_NEAREST)

def compute_normal_based_mapping(normal_map):
    """Compute texture mapping based on surface normals"""
    # Extract normal components
    nx = normal_map[:,:,0]
    ny = normal_map[:,:,1]
    nz = normal_map[:,:,2]

    # Compute tangent space basis vectors
    tangent = np.zeros_like(normal_map)
    tangent[:,:,0] = nz
    tangent[:,:,1] = 0
    tangent[:,:,2] = -nx
    tangent = tangent / (np.linalg.norm(tangent, axis=2, keepdims=True) + 1e-7)

    # Compute bitangent vectors
    bitangent = np.cross(normal_map, tangent)

    return tangent, bitangent

def apply_texture_mapping(frame, texture, normal_map, mask, clothing_type='upper'):
    """Apply texture with proper normal-based mapping"""
    h, w = frame.shape[:2]
    texture_h, texture_w = texture.shape[:2]

    # Find clothing region
    coords = np.where(mask > 0)
    if len(coords[0]) == 0:
        return frame

    y_min, y_max = np.min(coords[0]), np.max(coords[0])
    x_min, x_max = np.min(coords[1]), np.max(coords[1])

    # Get ROI
    roi_normal = normal_map[y_min:y_max, x_min:x_max]
    roi_mask = mask[y_min:y_max, x_min:x_max]

    # Compute normal-based mapping
    tangent, bitangent = compute_normal_based_mapping(roi_normal)

    # Create coordinate grid
    y_coords, x_coords = np.mgrid[0:roi_normal.shape[0], 0:roi_normal.shape[1]].astype(np.float32)
    coords = np.stack([x_coords, y_coords], axis=-1)

    # Apply normal-based deformation
    deform_strength = 0.1
    coords_3d = np.stack([
        coords[:,:,0] + tangent[:,:,0] * deform_strength,
        coords[:,:,1] + tangent[:,:,1] * deform_strength,
        np.ones_like(coords[:,:,0]) + tangent[:,:,2] * deform_strength
    ], axis=-1)

    # Project to texture space
    tex_x = ((coords_3d[:,:,0] / coords_3d[:,:,2]) / roi_normal.shape[1] * texture_w)
    tex_y = ((coords_3d[:,:,1] / coords_3d[:,:,2]) / roi_normal.shape[0] * texture_h)

    # Sample texture
    warped_texture = cv2.remap(texture,
                             np.clip(tex_x, 0, texture_w-1).astype(np.float32),
                             np.clip(tex_y, 0, texture_h-1).astype(np.float32),
                             cv2.INTER_LINEAR,
                             borderMode=cv2.BORDER_WRAP)

    # Safety check: ensure warped_texture has correct shape
    if len(warped_texture.shape) != 3 or warped_texture.shape[2] != 4:
        # Reconstruct with alpha channel
        if len(warped_texture.shape) == 2:
            # Grayscale
            gray = warped_texture
            bgr = cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([bgr, alpha])
        elif warped_texture.shape[2] == 3:
            # BGR without alpha
            alpha = np.ones(warped_texture.shape[:2], dtype=np.uint8) * 255
            warped_texture = np.dstack([warped_texture, alpha])

    # Calculate lighting - Different settings for upper vs lower clothing
    light_dir = np.array([0.0, 0.0, 1.0])

    if clothing_type == 'upper':
        # Upper clothing: brighter, preserves color better
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.5, 1.0)
        lighting = np.expand_dims(lighting, axis=2)

        # Apply lighting to texture color
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2)
        textured_rgb = warped_texture[:,:,:3].astype(np.float32) * lighting

    else:  # clothing_type == 'lower'
        # Lower clothing: improved color preservation with better lighting
        lighting = np.clip(np.sum(roi_normal * light_dir, axis=2), 0.5, 1.0)
        lighting = np.expand_dims(lighting, axis=2)

        # Apply lighting to texture RGB instead of alpha
        alpha = warped_texture[:,:,3] / 255.0 * roi_mask
        alpha = np.expand_dims(alpha, axis=2)
        textured_rgb = warped_texture[:,:,:3].astype(np.float32) * lighting

    # Blend with frame
    roi = frame[y_min:y_max, x_min:x_max].astype(np.float32)
    roi_new = roi * (1 - alpha) + textured_rgb * alpha
    frame[y_min:y_max, x_min:x_max] = np.clip(roi_new, 0, 255).astype(np.uint8)

    return frame

def process_frame(frame, normal_estimator, segmentation_estimator, upper_texture, lower_texture):
    """Process a single frame"""
    # Process the frame
    segmentation_map = segmentation_estimator(frame)
    normal_map = normal_estimator(frame)

    # Create masks for upper and lower clothing
    upper_mask = (segmentation_map == 22).astype(np.uint8)  # 22 = upper clothes
    lower_mask = (segmentation_map == 12).astype(np.uint8)  # 12 = lower clothes/pants

    result = frame.copy()

    # Apply texture mapping for upper clothing (shirt) if detected
    if np.any(upper_mask) and upper_texture is not None:
        result = apply_texture_mapping(result, upper_texture, normal_map, upper_mask, clothing_type='upper')

    # Apply texture mapping for lower clothing (pants) if detected
    if np.any(lower_mask) and lower_texture is not None:
        result = apply_texture_mapping(result, lower_texture, normal_map, lower_mask, clothing_type='lower')

    return result

def clear_memory():
    """Clear RAM and GPU memory"""
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()

def process_video(input_video_path, output_video_path, normal_estimator, segmentation_estimator,
                  upper_texture, lower_texture, skip_frames=0, start_time=None, end_time=None):
    """
    Process video file

    Args:
        start_time: Start time in seconds (None = from beginning)
        end_time: End time in seconds (None = until end)
    """
    # Open video
    cap = cv2.VideoCapture(input_video_path)
    if not cap.isOpened():
        print(f"Error: Could not open video from {input_video_path}")
        return False

    # Get video properties
    fps = int(cap.get(cv2.CAP_PROP_FPS))
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    total_duration = total_frames / fps

    print(f"Video properties: {width}x{height} @ {fps}fps")
    print(f"Total duration: {total_duration:.2f} seconds ({total_frames} frames)")

    # Calculate start and end frames
    start_frame = 0
    end_frame = total_frames

    if start_time is not None:
        start_frame = int(start_time * fps)
        start_frame = max(0, min(start_frame, total_frames - 1))
        print(f"Start time: {start_time:.2f}s (frame {start_frame})")

    if end_time is not None:
        end_frame = int(end_time * fps)
        end_frame = max(start_frame + 1, min(end_frame, total_frames))
        print(f"End time: {end_time:.2f}s (frame {end_frame})")

    frames_to_process = end_frame - start_frame
    print(f"Processing {frames_to_process} frames ({frames_to_process/fps:.2f} seconds)")

    # Set video to start position
    if start_frame > 0:
        cap.set(cv2.CAP_PROP_POS_FRAMES, start_frame)

    # Define codec and create VideoWriter
    fourcc = cv2.VideoWriter_fourcc(*'mp4v')
    out = cv2.VideoWriter(output_video_path, fourcc, fps, (width, height))

    frame_count = start_frame
    processed_count = 0

    print("\nProcessing video...")

    try:
        while frame_count < end_frame:
            ret, frame = cap.read()
            if not ret:
                break

            # Skip frames if specified
            if skip_frames > 0 and (frame_count - start_frame) % (skip_frames + 1) != 0:
                out.write(frame)  # Write original frame
                frame_count += 1
                continue

            # Process frame
            try:
                result = process_frame(frame, normal_estimator, segmentation_estimator,
                                     upper_texture, lower_texture)
                out.write(result)
                processed_count += 1

                # Progress indicator
                if processed_count % 10 == 0:
                    progress = ((frame_count - start_frame) / frames_to_process) * 100
                    current_time = frame_count / fps
                    print(f"Progress: {progress:.1f}% - Time: {current_time:.2f}s / {end_time if end_time else total_duration:.2f}s")

            except Exception as e:
                print(f"Error processing frame {frame_count}: {e}")
                out.write(frame)  # Write original frame on error

            frame_count += 1

            # Clear memory periodically
            if processed_count % 50 == 0:
                clear_memory()

    finally:
        cap.release()
        out.release()
        print(f"\n✓ Video processing complete: {processed_count} frames processed")
        print(f"✓ Output saved to: {output_video_path}")

    return True

def main():
    # Clear memory before processing
    print("Clearing memory before processing...")
    clear_memory()
    print("✓ Memory cleared\n")

    # ===== CONFIGURATION =====
    # Set MODE to 'image' or 'video'
    MODE = 'video'  # Change to 'video' for video processing

    # Define paths
    if MODE == 'image':
        input_path = '/content/drive/MyDrive/ATL-Project/test.jpg'
        output_path = '/content/drive/MyDrive/ATL-Project/result.png'
    else:  # video
        input_path = '/content/drive/MyDrive/ATL-Project/test_video_5.mp4'
        output_path = '/content/drive/MyDrive/ATL-Project/result_video_5.mp4'

        # Video processing options
        SKIP_FRAMES = 0     # Process every frame. Set to 1 for every other frame, 2 for every 3rd, etc.
        START_TIME = 2   # Start time in seconds (None = from beginning, e.g., 5.0 for 5 seconds)
        END_TIME = 4     # End time in seconds (None = until end, e.g., 15.0 for 15 seconds)

        # Examples:
        # START_TIME = 0      # Start from beginning
        # END_TIME = 2      # Process first 10 seconds
        # START_TIME = 5.5    # Start at 5.5 seconds
        # END_TIME = 20.5     # End at 20.5 seconds
        # START_TIME = 10
        # END_TIME = 30       # Process from 10s to 30s (20 seconds total)

    # Texture paths
    upper_texture_path = '/content/drive/MyDrive/ATL-Project/red-fabric-texture.jpg'
    lower_texture_path = '/content/drive/MyDrive/ATL-Project/black_cotton.jpg'

    # Initialize models
    print("Loading models...")
    normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
    segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)
    print("✓ Models loaded\n")

    # Load textures
    upper_texture = None
    lower_texture = None

    try:
        upper_texture = load_texture(upper_texture_path)
        print("✓ Upper texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")

    try:
        lower_texture = load_texture(lower_texture_path)
        print("✓ Lower texture loaded successfully")
    except ValueError as e:
        print(f"Warning: {e}")

    print()

    # Process based on mode
    if MODE == 'image':
        # ===== IMAGE PROCESSING =====
        print("=== Processing Image ===")
        frame = cv2.imread(input_path)
        if frame is None:
            print(f"Error: Could not load image from {input_path}")
            return

        result = process_frame(frame, normal_estimator, segmentation_estimator,
                             upper_texture, lower_texture)

        # Save result
        if not np.array_equal(result, frame):
            cv2.imwrite(output_path, result)
            print(f"✓ Result saved to {output_path}")
        else:
            print("No clothing detected or no textures applied")

    else:  # video
        # ===== VIDEO PROCESSING =====
        print("=== Processing Video ===")
        success = process_video(input_path, output_path, normal_estimator, segmentation_estimator,
                              upper_texture, lower_texture,
                              skip_frames=SKIP_FRAMES,
                              start_time=START_TIME,
                              end_time=END_TIME)

        if not success:
            print("Video processing failed")
            return

    # Clear memory after processing
    print("\nClearing memory after processing...")
    clear_memory()
    print("✓ Processing complete")

if __name__ == "__main__":
    main()
```

```python
import torch
import cv2
import numpy as np
from pathlib import Path
from sapiens_inference.normal import SapiensNormal, SapiensNormalType
from sapiens_inference.segmentation import SapiensSegmentation, SapiensSegmentationType

# Sapiens Segmentation Class Labels
SEGMENTATION_CLASSES = {
    0: 'Background',
    1: 'Torso-Skin',
    2: 'Left-Hand',
    3: 'Right-Hand',
    4: 'Left-Foot',
    5: 'Right-Foot',
    6: 'Upper-Clothing',
    7: 'Lower-Clothing',
    8: 'Hair',
    9: 'Left-Shoe',
    10: 'Right-Shoe',
    11: 'Socks',
    12: 'Headwear',
    13: 'Scarf',
    14: 'Coat',
    15: 'Dress',
    16: 'Face',
    17: 'Left-Arm',
    18: 'Right-Arm',
    19: 'Left-Leg',
    20: 'Right-Leg',
    21: 'Left-Shoe',
    22: 'Right-Shoe',
}

class TShirtTextureTransferFixed:
    def __init__(self, texture_path, uv_scale=0.5, blend_strength=0.5):
        """
        Initialize the t-shirt texture transfer system.

        Args:
            texture_path (str): Path to the texture image with alpha channel
            uv_scale (float): Scale factor for UV coordinates
            blend_strength (float): Strength of texture blending (0-1)
        """
        self.uv_scale = uv_scale
        self.blend_strength = blend_strength

        # Load texture
        self.texture = cv2.imread(texture_path, cv2.IMREAD_UNCHANGED)
        if self.texture is None:
            raise ValueError(f'Failed to load texture from {texture_path}')

        # Separate RGB and alpha channels
        if len(self.texture.shape) == 3 and self.texture.shape[2] == 4:
            self.texture_rgb = self.texture[..., :3]
            self.texture_alpha = self.texture[..., 3] / 255.0
        else:
            print("Warning: Texture doesn't have alpha channel, creating one...")
            if len(self.texture.shape) == 2:
                self.texture_rgb = cv2.cvtColor(self.texture, cv2.COLOR_GRAY2BGR)
            else:
                self.texture_rgb = self.texture[..., :3]
            self.texture_alpha = np.ones(self.texture_rgb.shape[:2], dtype=np.float32)

        # Initialize models
        print("Initializing Sapiens models...")
        self.normal_estimator = SapiensNormal(SapiensNormalType.NORMAL_06B)
        self.segmentation_estimator = SapiensSegmentation(SapiensSegmentationType.SEGMENTATION_06B)
        print("Models initialized successfully!")

        # For temporal smoothing in videos
        self.previous_normal_map = None

    def visualize_segmentation(self, frame, segmentation_map, output_path):
        """
        Create and save a visualization of the segmentation map with class labels.

        Args:
            frame (np.ndarray): Original frame
            segmentation_map (np.ndarray): Segmentation output
            output_path (str): Path to save visualization
        """
        # Ensure segmentation_map is integer type
        if segmentation_map.dtype in [np.float32, np.float64]:
            segmentation_map = np.round(segmentation_map).astype(np.int32)
        else:
            segmentation_map = segmentation_map.astype(np.int32)

        # Create color-coded visualization
        h, w = segmentation_map.shape
        vis = np.zeros((h, w, 3), dtype=np.uint8)

        # Generate distinct colors for each class
        np.random.seed(42)
        colors = {}
        for class_id in range(23):
            colors[class_id] = tuple(np.random.randint(0, 255, 3).tolist())

        # Special colors for important classes
        colors[6] = (255, 0, 0)    # Upper-Clothing - Blue
        colors[14] = (0, 255, 255)  # Coat - Cyan
        colors[22] = (0, 255, 0)    # Class 22 - Green

        # Get unique classes
        unique_classes = np.unique(segmentation_map)

        # Apply colors
        for class_id in unique_classes:
            class_id_int = int(class_id)
            mask = segmentation_map == class_id
            if class_id_int in colors:
                vis[mask] = colors[class_id_int]

        # Create overlay
        overlay = cv2.addWeighted(frame, 0.5, vis, 0.5, 0)

        # Add legend
        legend_height = 30
        legend_width = 350
        detected_classes = sorted([int(c) for c in unique_classes])
        legend = np.zeros((len(detected_classes) * legend_height + 10, legend_width, 3), dtype=np.uint8)

        for i, class_id in enumerate(detected_classes):
            # Draw color box
            cv2.rectangle(legend,
                         (10, i * legend_height + 5),
                         (40, (i + 1) * legend_height - 5),
                         colors.get(class_id, (128, 128, 128)), -1)

            # Add text
            class_name = SEGMENTATION_CLASSES.get(class_id, f'Unknown-{class_id}')
            pixel_count = int(np.sum(segmentation_map == class_id))
            text = f"{class_id}: {class_name} ({pixel_count}px)"

            cv2.putText(legend, text, (50, i * legend_height + 20),
                       cv2.FONT_HERSHEY_SIMPLEX, 0.4, (255, 255, 255), 1)

        # Combine overlay and legend
        if legend.shape[0] > overlay.shape[0]:
            legend = legend[:overlay.shape[0], :, :]
        else:
            padding = overlay.shape[0] - legend.shape[0]
            legend = np.vstack([legend, np.zeros((padding, legend_width, 3), dtype=np.uint8)])

        result = np.hstack([overlay, legend])

        cv2.imwrite(output_path, result)
        print(f"Segmentation visualization saved to: {output_path}")

        # Save masks for clothing classes
        for class_id in [6, 14, 22]:
            if class_id in detected_classes:
                class_mask = (segmentation_map == class_id).astype(np.uint8) * 255
                mask_path = output_path.replace('.', f'_class{class_id}_mask.')
                cv2.imwrite(mask_path, class_mask)
                print(f"Class {class_id} mask saved to: {mask_path}")

        return detected_classes

    def apply_texture_to_normals(self, normal_map):
        """
        Apply texture based on normal map with UV mapping.

        Args:
            normal_map (np.ndarray): Normal map from Sapiens

        Returns:
            tuple: (textured_rgb, textured_alpha)
        """
        # Normalize normal map to [0, 1] for UV mapping
        normalized_normals = (normal_map + 1) / 2
        normalized_normals = np.clip(normalized_normals, 0, 1)

        # Use x,y components as UV coordinates
        uv_coords = normalized_normals[:, :, :2]

        # Convert to texture pixel indices with scaling
        tex_h, tex_w = self.texture_rgb.shape[:2]
        u_indices = (uv_coords[:, :, 0] * tex_w * self.uv_scale).astype(np.int32)
        v_indices = (uv_coords[:, :, 1] * tex_h * self.uv_scale).astype(np.int32)

        # Wrap coordinates for tiling effect
        u_indices = u_indices % tex_w
        v_indices = v_indices % tex_h

        # Sample texture
        sampled_rgb = self.texture_rgb[v_indices, u_indices]
        sampled_alpha = self.texture_alpha[v_indices, u_indices]

        return sampled_rgb, sampled_alpha

    def process_frame(self, frame, target_classes=[6, 22], use_temporal_smoothing=False,
                     save_segmentation=False, output_prefix=''):
        """
        Process a single frame to apply texture to specified clothing classes.

        Args:
            frame (np.ndarray): Input frame (BGR)
            target_classes (list): List of segmentation class IDs to apply texture to
            use_temporal_smoothing (bool): Whether to apply temporal smoothing
            save_segmentation (bool): Whether to save segmentation visualization
            output_prefix (str): Prefix for output files

        Returns:
            tuple: (processed_frame, detected_classes)
        """
        try:
            # Get segmentation mask
            segmentation_map = self.segmentation_estimator(frame)

            # Convert to integer if needed
            if segmentation_map.dtype in [np.float32, np.float64]:
                segmentation_map = np.round(segmentation_map).astype(np.int32)
            else:
                segmentation_map = segmentation_map.astype(np.int32)

            # Visualize segmentation if requested
            detected_classes = []
            if save_segmentation:
                seg_path = f'{output_prefix}_segmentation.jpg'
                detected_classes = self.visualize_segmentation(frame, segmentation_map, seg_path)

            # Create combined mask for all target classes
            combined_mask = np.zeros(segmentation_map.shape, dtype=np.uint8)
            for class_id in target_classes:
                mask_for_class = (segmentation_map == class_id)
                combined_mask = np.logical_or(combined_mask, mask_for_class)
            combined_mask = combined_mask.astype(np.uint8)

            if not np.any(combined_mask):
                print(f"Warning: No target clothing classes {target_classes} detected!")
                if not detected_classes:
                    detected_classes = list(np.unique(segmentation_map))
                print(f"Available classes: {detected_classes}")
                return frame, detected_classes

            print(f"Found {np.sum(combined_mask)} pixels in target classes {target_classes}")

            # Estimate normal map
            current_normal_map = self.normal_estimator(frame)

            # Apply Gaussian blur for smoother results
            blurred_normal_map = cv2.GaussianBlur(current_normal_map, (7, 7), sigmaX=2, sigmaY=2)

            # Temporal smoothing for videos
            if use_temporal_smoothing and self.previous_normal_map is not None:
                normal_map = cv2.addWeighted(blurred_normal_map, 0.7, self.previous_normal_map, 0.3, 0)
            else:
                normal_map = blurred_normal_map

            self.previous_normal_map = normal_map.copy()

            # Apply texture using normal map
            textured_rgb, textured_alpha = self.apply_texture_to_normals(normal_map)

            # Apply mask to texture (FULL COVERAGE)
            textured_rgb = textured_rgb * combined_mask[:, :, None]
            textured_alpha = textured_alpha * combined_mask

            # Blend texture with original frame
            result = frame.copy().astype(np.float32)
            mask_indices = combined_mask == 1

            # Alpha blending with configurable strength
            alpha_blend = self.blend_strength * textured_alpha[mask_indices]
            alpha_blend = np.clip(alpha_blend, 0, 1)

            # Ensure shapes match
            alpha_expanded = alpha_blend[:, np.newaxis]

            result[mask_indices] = (
                textured_rgb[mask_indices].astype(np.float32) * alpha_expanded +
                frame[mask_indices].astype(np.float32) * (1 - alpha_expanded)
            )

            result = np.clip(result, 0, 255).astype(np.uint8)

            return result, detected_classes

        except Exception as e:
            print(f"Error in process_frame: {e}")
            import traceback
            traceback.print_exc()
            raise

    def process_image(self, input_path, output_path, target_classes=[6, 22],
                     save_segmentation=True):
        """
        Process a single image.

        Args:
            input_path (str): Path to input image
            output_path (str): Path to save output image
            target_classes (list): List of class IDs to apply texture to
            save_segmentation (bool): Whether to save segmentation visualization
        """
        print(f"Processing image: {input_path}")

        frame = cv2.imread(input_path)
        if frame is None:
            raise ValueError(f"Failed to load image from {input_path}")

        print(f"Image loaded: {frame.shape}")

        # Create output prefix for segmentation files
        output_prefix = str(Path(output_path).with_suffix(''))

        result, detected_classes = self.process_frame(
            frame,
            target_classes=target_classes,
            use_temporal_smoothing=False,
            save_segmentation=save_segmentation,
            output_prefix=output_prefix
        )

        cv2.imwrite(output_path, result)
        print(f"Saved result to: {output_path}")

        if detected_classes:
            print(f"\nDetected classes: {detected_classes}")
            print("\nClass mapping:")
            for class_id in detected_classes:
                class_id = int(class_id)
                class_name = SEGMENTATION_CLASSES.get(class_id, f'Unknown-{class_id}')
                print(f"  {class_id}: {class_name}")

        return result, detected_classes

    def process_video(self, input_path, output_path, target_classes=[6, 22],
                     show_progress=True):
        """
        Process a video file.

        Args:
            input_path (str): Path to input video
            output_path (str): Path to save output video
            target_classes (list): List of class IDs to apply texture to
            show_progress (bool): Whether to display progress
        """
        print(f"Processing video: {input_path}")

        # Open video
        cap = cv2.VideoCapture(input_path)
        if not cap.isOpened():
            raise ValueError(f"Failed to open video: {input_path}")

        # Get video properties
        fps = int(cap.get(cv2.CAP_PROP_FPS))
        width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

        # Create video writer
        fourcc = cv2.VideoWriter_fourcc(*'mp4v')
        out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

        # Reset temporal smoothing
        self.previous_normal_map = None

        frame_count = 0
        save_seg_first_frame = True

        while True:
            ret, frame = cap.read()
            if not ret:
                break

            # Save segmentation for first frame
            save_seg = save_seg_first_frame
            output_prefix = str(Path(output_path).with_suffix('')) + '_frame0' if save_seg else ''

            # Process frame
            result, _ = self.process_frame(
                frame,
                target_classes=target_classes,
                use_temporal_smoothing=True,
                save_segmentation=save_seg,
                output_prefix=output_prefix
            )
            out.write(result)

            save_seg_first_frame = False
            frame_count += 1

            if show_progress and frame_count % 10 == 0:
                progress = (frame_count / total_frames) * 100
                print(f"Progress: {progress:.1f}% ({frame_count}/{total_frames} frames)")

        # Cleanup
        cap.release()
        out.release()

        print(f"Video processing complete! Saved to: {output_path}")


def main():
    """
    Example usage - UPDATE THE PATHS!
    """
    # Configuration
    TEXTURE_PATH = '/content/drive/MyDrive/ATL-Project/texture.png'
    INPUT_IMAGE = '/content/drive/MyDrive/ATL-Project/demo.jpg'
    OUTPUT_IMAGE = '/content/drive/MyDrive/ATL-Project/output_full_shirt.jpg'

    # Try different class combinations
    # Most common: [6] for upper clothing, [22] for some jerseys
    # Safe bet: [6, 22] covers most cases
    TARGET_CLASSES = [6, 22]  # Try [6], [22], or [6, 22]

    # Advanced settings
    UV_SCALE = 0.5          # 0.3-0.8: smaller = finer texture details
    BLEND_STRENGTH = 0.9   # 0.3-0.7: higher = more visible texture

    try:
        # Initialize processor
        processor = TShirtTextureTransferFixed(
            texture_path=TEXTURE_PATH,
            uv_scale=UV_SCALE,
            blend_strength=BLEND_STRENGTH
        )

        # Process image with full segmentation visualization
        result, detected_classes = processor.process_image(
            INPUT_IMAGE,
            OUTPUT_IMAGE,
            target_classes=TARGET_CLASSES,
            save_segmentation=True  # Creates segmentation maps
        )

        print("\n" + "="*60)
        print("PROCESSING COMPLETE!")
        print("="*60)
        print(f"\nOutput files:")
        print(f"1. {OUTPUT_IMAGE}")
        print(f"2. {OUTPUT_IMAGE.replace('.jpg', '_segmentation.jpg')}")

        if detected_classes:
            clothing_classes = [c for c in detected_classes if c in [6, 14, 22]]
            if clothing_classes:
                print(f"\nDetected clothing classes: {clothing_classes}")
                print(f"If the result doesn't look right, try:")
                print(f"  TARGET_CLASSES = {clothing_classes}")

    except Exception as e:
        print(f"\nError: {e}")
        import traceback
        traceback.print_exc()


if __name__ == "__main__":
    main()
```
